```python
import random

N = 4

# Calculate the number of attacking queen pairs
def heuristic(board):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


# Print the chessboard
def print_board(board):
    for row in range(N):
        for col in range(N):
            if board[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()


# Hill Climbing algorithm
def hill_climbing():

    # Random initial state (one queen per column)
    board = [random.randint(0, N - 1) for _ in range(N)]

    step = 1

    while True:

        print("\nStep:", step)
        print("Current State:", board)
        print("Heuristic Value:", heuristic(board))
        print_board(board)

        current_h = heuristic(board)

        # Solution found
        if current_h == 0:
            print("\nSolution Found!")
            return board

        best_board = board[:]
        best_h = current_h

        # Generate neighboring states
        for col in range(N):
            for row in range(N):

                if board[col] != row:
                    neighbor = board[:]
                    neighbor[col] = row

                    h = heuristic(neighbor)

                    print(
                        "Move queen in column",
                        col + 1,
                        "to row",
                        row + 1,
                        "=> Heuristic:",
                        h
                    )

                    if h < best_h:
                        best_h = h
                        best_board = neighbor

        # No better neighbor found
        if best_h >= current_h:
            print("\nLocal Maximum / Plateau reached!")
            print("Restarting with a new random state...\n")
            board = [random.randint(0, N - 1) for _ in range(N)]
        else:
            board = best_board

        step += 1


# Run the algorithm
solution = hill_climbing()

print("\nFinal Solution:", solution)
print("\nFinal Chessboard:")
print_board(solution)
```
