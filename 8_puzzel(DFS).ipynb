```python
GOAL = "123456780"


# =========================================================
# CHECK WHETHER PUZZLE IS SOLVABLE
# =========================================================

def is_solvable(state):
    nums = [int(x) for x in state if x != '0']

    inversions = 0

    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] > nums[j]:
                inversions += 1

    return inversions % 2 == 0


# =========================================================
# GENERATE NEIGHBORS
# =========================================================

def get_neighbors(state):

    neighbors = []

    zero = state.index('0')

    row = zero // 3
    col = zero % 3

    moves = [
        (1, 0),     # Down
        (0, 1),     # Right
        (-1, 0),    # Up
        (0, -1)     # Left
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_pos = new_row * 3 + new_col

            temp = list(state)

            temp[zero], temp[new_pos] = \
                temp[new_pos], temp[zero]

            neighbors.append("".join(temp))

    return neighbors


# =========================================================
# PRINT STATE
# =========================================================

def print_state(state):

    for i in range(0, 9, 3):
        print(" ".join(
            "_" if x == '0' else x
            for x in state[i:i+3]
        ))

    print()


# =========================================================
# DFS
# =========================================================

def dfs(start, max_depth=30):

    stack = [(start, [start])]

    visited = set()

    while stack:

        state, path = stack.pop()

        if state == GOAL:
            return path

        if len(path) - 1 >= max_depth:
            continue

        if state in visited:
            continue

        visited.add(state)

        for neighbor in get_neighbors(state):

            if neighbor not in visited:

                stack.append(
                    (neighbor, path + [neighbor])
                )

    return None


# =========================================================
# DEPTH LIMITED SEARCH
# =========================================================

def depth_limited_search(
        state,
        depth,
        path,
        visited):

    if state == GOAL:
        return path

    if depth == 0:
        return None

    visited.add(state)

    for neighbor in get_neighbors(state):

        if neighbor not in visited:

            result = depth_limited_search(
                neighbor,
                depth - 1,
                path + [neighbor],
                visited
            )

            if result is not None:
                return result

    # Backtracking
    visited.remove(state)

    return None


# =========================================================
# ITERATIVE DEEPENING SEARCH
# =========================================================

def ids(start, max_depth=30):

    for depth in range(max_depth + 1):

        print("IDS depth:", depth)

        visited = set()

        result = depth_limited_search(
            start,
            depth,
            [start],
            visited
        )

        if result is not None:
            return result

    return None


# =========================================================
# MAIN
# =========================================================

start = "123560478"

print("Initial State:")
print_state(start)


# ---------------------------------------------------------
# SOLVABILITY CHECK
# ---------------------------------------------------------

if not is_solvable(start):

    print("Puzzle is NOT solvable.")
    exit()


# =========================================================
# DFS
# =========================================================

print("========== DFS ==========")

solution = dfs(start, max_depth=30)

if solution is not None:

    print("Solution Found!")
    print("Number of moves:", len(solution) - 1)
    print()

    for state in solution:
        print_state(state)

else:

    print("No solution found within depth limit.")


# =========================================================
# IDS
# =========================================================

print("========== IDS ==========")

solution = ids(start, max_depth=30)

if solution is not None:

    print("Solution Found!")
    print("Number of moves:", len(solution) - 1)
    print()

    for state in solution:
        print_state(state)

else:

    print("No solution found within depth limit.")
```
