```python

N = 4

# Calculate attacking queen pairs
def heuristic(board):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


# Print chessboard
def print_board(board):
    for row in range(N):
        for col in range(N):
            if board[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()


# Generate exactly 6 neighbors by swapping two columns
def generate_neighbors(board):
    neighbors = []

    for i in range(N):
        for j in range(i + 1, N):
            neighbor = board[:]

            # Swap queen positions
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

            neighbors.append(neighbor)

    return neighbors


# Hill Climbing Algorithm
def hill_climbing():

    # Initial state
    board = [3,1,2,0]

    step = 1

    while True:

        current_h = heuristic(board)

        print("\n========== STEP", step, "==========")
        print("Current State:", board)
        print("Heuristic:", current_h)
        print_board(board)

        if current_h == 0:
            print("\nSolution Found!")
            return board

        # Generate 6 neighbors
        neighbors = generate_neighbors(board)

        print("\nAll 6 Neighbors:")

        for neighbor in neighbors:
            print(neighbor, "Heuristic:", heuristic(neighbor))

        # Find minimum heuristic
        min_h = min(heuristic(n) for n in neighbors)

        # Select neighbors with minimum heuristic
        best_neighbors = [
            n for n in neighbors
            if heuristic(n) == min_h
        ]

        # Lexicographically smallest neighbor
        best_neighbor = min(best_neighbors)

        print("\nMinimum Heuristic:", min_h)
        print("Selected Neighbor:", best_neighbor)

        # Check improvement
        if min_h >= current_h:
            print("\nNo better neighbor found.")
            print("Algorithm stopped.")
            return None

        board = best_neighbor
        step += 1


# Run algorithm
solution = hill_climbing()

if solution:
    print("\nFinal Solution:", solution)
    print("\nFinal Board:")
    print_board(solution)
```
