```python
import random
import math

def get_h_score(board):
    """
    Calculates the heuristic cost (h-score): the number of mutually attacking pairs of queens.
    board: list of integers where index=column and value=row.
    """
    n = len(board)
    attacks = 0
    for i in range(n):
        for j in range(i + 1, n):
            # Same row conflict or diagonal conflict
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                attacks += 1
    return attacks

def get_random_neighbor(board):
    """
    Generates a single random neighbor by picking a random column
    and moving its queen to a random different row.
    """
    n = len(board)
    neighbor = list(board)

    col = random.randint(0, n - 1)
    current_row = board[col]

    # Pick a new row that is different from the current row
    new_row = random.randint(0, n - 1)
    while new_row == current_row:
        new_row = random.randint(0, n - 1)

    neighbor[col] = new_row
    return neighbor

def simulated_annealing_8_queens(initial_state=None):
    """
    Solves the 8-Queens problem using the Simulated Annealing metaheuristic.
    Accepts an optional initial_state. If the search cools down without finding a solution,
    it falls back to a clean random restart loop.
    """
    n = 8

    # Use the provided initial state if present; otherwise, generate a random one
    if initial_state is not None:
        current_state = list(initial_state)
    else:
        current_state = [random.randint(0, n - 1) for _ in range(n)]

    current_h = get_h_score(current_state)

    # Temperature schedule parameters
    T = 100.0          # Initial temperature
    cooling_rate = 0.95 # Geometric cooling factor
    min_T = 0.001      # Threshold to stop cooling loop
    max_steps_per_T = 100 # Internal iterations before cooling down

    # Main schedule loop
    while T > min_T:
        if current_h == 0:
            return current_state  # Global optimum found

        for _ in range(max_steps_per_T):
            if current_h == 0:
                return current_state

            # Pick a single random neighbor state
            neighbor = get_random_neighbor(current_state)
            neighbor_h = get_h_score(neighbor)

            # Compute the delta change in evaluation cost
            delta_E = neighbor_h - current_h

            # Accept if better, or accept probabilistically if worse
            if delta_E < 0 or random.random() < math.exp(-delta_E / T):
                current_state = neighbor
                current_h = neighbor_h

        # Cool down the system state
        T *= cooling_rate

    # If system cooled down completely without reaching 0 conflicts, retry using random restarts
    return simulated_annealing_8_queens(initial_state=None)

def print_board(board):
    """Utility to print the board layout cleanly."""
    n = len(board)
    for row in range(n):
        line = ""
        for col in range(n):
            line += " Q " if board[col] == row else " . "
        print(line)

# Execution block
if __name__ == "__main__":
    # Define your custom initial state here (example for 8-queens)
    custom_initial_state = [1,4,8,5,2,3,6,7]

    print(f"Finding a solution for 8-Queens starting from initial state: {custom_initial_state}...\n")
    solution = simulated_annealing_8_queens(initial_state=custom_initial_state)

    if solution:
        print(f"Solution state vector (rows per column): {solution}\n")
        print_board(solution)
```
